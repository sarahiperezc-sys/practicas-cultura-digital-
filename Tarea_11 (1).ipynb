## Parte 3: Condicionales en Python

### Ejercicio 1
Pide la hora (0-23). Si es menor que 12 muestra 'Es mañana', si no 'Es tarde o noche'.


```python
hora=int(input("ingrese la hora:"))
if hora<12:
    print("es mañana")
else:
    print("es tarde o noche")
```

### Ejercicio 2
Pide dos números y muestra cuál es mayor o si son iguales.


```python
var1=int(input("ingrese el primer numero:"))
var2=int(input("ingrese el segundo numero:"))
if var1>var2:
  print("el primer numero es mayor")
elif var2>var1:
  print("el segundo numero es mayor")
else:
  print("los numeros son iguales")
```

### Ejercicio 3
Pide un número entero y muestra si es par o impar.


```python
numero=int(input("ingrese un numero entero"))
if numero%2==0:
  print("el numero es par.")
else:
  print("el numero es impar.")
```
